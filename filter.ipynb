#Filtering data -- using Logical(&, |) Operators

```python
import pandas as pd
```

```python
b=pd.read_csv("sample_data.csv")
b
```

```python
#Filtering
b[b["marks"] > 75]
```

```python
b[(b["marks"]>75) & (b["age"]>20)]
```

```python
b[(b["marks"]> 90) | (b["salary"]>55000)]
```

```python
b[(b["marks"]>80) & (b["age"]>20) & (b["salary"]>50000)]
```

```python
b[(b["marks"]>85) | (b["age"]>24) & (b["salary"]> 60000) ]
```
